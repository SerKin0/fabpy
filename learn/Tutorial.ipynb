### Среднеквадратичное отклонение
Статистическая характеристика распределения случайной величины, показывающая среднюю степень разброса значений относительно математического ожидания. 

$$S_t = \sqrt{\frac{\sum_{i=1}^n (\overline{t} - t_i)^2}{n}}$$

Расмотрим на примере:

```python
from fabpy import Values

# Список эксперементальных данных
data_t = [10.5, 10.7, 10.6]

# Создание переменной времени с соответсующими данными
t = Values(name='t', values=data_t, unit=r'с', delta=0.1)

# Вывод полного значения 
print(t.standard_deviation.value)

# Округленное значение. По умолчанию округляется по заданному roundoff
# При необходимости количество знаков можно указать в параметре round_value()
print(t.standard_deviation.round_value())

# Вывод получения формулы 
print(t.standard_deviation.latex())
```

```python
data_t = [10.5, 10.7, 10.6]

# Создание переменной времени с соответсующими данными
t = Values(name='t', values=data_t, unit=r'с', delta=0.1)

# Полная формула
# Выводим без фрагмента общей формулы (general)
print(t.absolute_error.latex(
    print_name = True,
    print_general = False,
    print_values = True,
    print_result = True))

# Краткая форма
# Выводим только название переменной и ее значение
print(t.absolute_error.latex(
    print_general = False,
    print_values = False
)
)
```
